```python
import pandas as pd


# ---------------------------------------------------------
# Q1: Build Personalized Knowledge Base
# Roll Number: 1024170001
# Last two digits = 0, 1
# 0 % 3 = billing
# 1 % 3 = account
# ---------------------------------------------------------

def q1_create_knowledge_base():
    fixed_entries = [
        {
            "question": "what is the annual fee",
            "answer": "The annual fee is Rs 500.",
            "keywords": "fee cost price charge",
            "category": "billing"
        },
        {
            "question": "how to reset password",
            "answer": "Go to Settings > Reset Password.",
            "keywords": "password reset login",
            "category": "account"
        },
        {
            "question": "what are your working hours",
            "answer": "We are open 9 AM to 5 PM.",
            "keywords": "hours timing open time",
            "category": "general"
        },
        {
            "question": "how can i pay the fee",
            "answer": "You can pay via UPI, card, or net banking.",
            "keywords": "pay payment upi fee",
            "category": "billing"
        }
    ]

    # Last digit = 0 -> billing
    personalized_1 = {
        "question": "how can i check my payment status",
        "answer": "You can check your payment status from the billing section.",
        "keywords": "payment status billing transaction",
        "category": "billing"
    }

    # Last digit = 1 -> account
    personalized_2 = {
        "question": "how do i update my registered mobile number",
        "answer": "Go to Account Settings and update your registered mobile number.",
        "keywords": "mobile number update account",
        "category": "account"
    }

    entries = fixed_entries + [personalized_1, personalized_2]

    df = pd.DataFrame(entries)

    print("\n========== Q1: FINAL KNOWLEDGE BASE ==========")
    print(df)

    return df


# ---------------------------------------------------------
# Q2: Generate and Score a Hypothesis
# ---------------------------------------------------------

def score_query(query, df):
    query_words = set(query.lower().split())

    results = []

    for index, row in df.iterrows():

        # Combine question and keywords
        text = (
            str(row["question"]).lower()
            + " "
            + str(row["keywords"]).lower()
        )

        text_words = set(text.split())

        # Number of matching words
        score = len(query_words.intersection(text_words))

        if score > 0:
            results.append({
                "index": index,
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "score": score
            })

    # Sort by confidence score
    results.sort(key=lambda x: x["score"], reverse=True)

    return results


def q2_scoring(df):
    print("\n========== Q2: QUERY SCORING ==========")

    query = input("Enter your query: ")

    results = score_query(query, df)

    if not results:
        print("No matching FAQ found.")
        return

    print("\nMatching entries:")

    for result in results:
        print(
            f"Score: {result['score']} | "
            f"Category: {result['category']} | "
            f"Question: {result['question']}"
        )


# ---------------------------------------------------------
# Q3: same_category(category_name, df)
# ---------------------------------------------------------

def same_category(category_name, df):
    return df[df["category"].str.lower() == category_name.lower()]


def q3_category(df):
    print("\n========== Q3: SAME CATEGORY ==========")

    # Category of one personalized entry
    category = "billing"

    result = same_category(category, df)

    print(f"Questions belonging to category: {category}")
    print(result[["question", "category"]])


# ---------------------------------------------------------
# Q4: Add a new keyword and save CSV
# ---------------------------------------------------------

def q4_add_keyword(df):
    print("\n========== Q4: ADD KEYWORD ==========")

    # Pick one entry
    index = 0

    print("Selected entry:")
    print(df.loc[index])

    new_keyword = input("\nEnter a new keyword: ")

    # Add keyword
    current_keywords = str(df.loc[index, "keywords"])

    if new_keyword not in current_keywords.split():
        df.loc[index, "keywords"] = current_keywords + " " + new_keyword

    filename = "1024170001_faq_data.csv"

    df.to_csv(filename, index=False)

    print(f"\nUpdated DataFrame saved as {filename}")

    print("\nUpdated entry:")
    print(df.loc[index])

    return df


# ---------------------------------------------------------
# Q5: GroupBy category
# ---------------------------------------------------------

def q5_groupby(df):
    print("\n========== Q5: FAQ COUNT BY CATEGORY ==========")

    category_count = df.groupby("category").size()

    print(category_count)


# ---------------------------------------------------------
# Q6: Handle Ties
# ---------------------------------------------------------

def q6_tie_handling(df):
    print("\n========== Q6: TIE HANDLING ==========")

    # ---------------------------------------------
    # Query that produces a tie
    # ---------------------------------------------

    query1 = "fee"

    results1 = score_query(query1, df)

    print("\nQuery 1:", query1)

    if results1:
        highest_score = results1[0]["score"]

        top_matches = [
            result for result in results1
            if result["score"] == highest_score
        ]

        if len(top_matches) > 1:
            print("TIE FOUND! All highest-scoring entries:")

            for result in top_matches:
                print(
                    f"Score: {result['score']} | "
                    f"Question: {result['question']} | "
                    f"Category: {result['category']}"
                )

        else:
            print("Best matching entry:")
            print(top_matches[0]["question"])

    # ---------------------------------------------
    # Query that does NOT produce a tie
    # ---------------------------------------------

    query2 = "password reset"

    results2 = score_query(query2, df)

    print("\nQuery 2:", query2)

    if results2:
        highest_score = results2[0]["score"]

        top_matches = [
            result for result in results2
            if result["score"] == highest_score
        ]

        if len(top_matches) > 1:
            print("TIE FOUND!")

            for result in top_matches:
                print(
                    f"Score: {result['score']} | "
                    f"Question: {result['question']}"
                )

        else:
            print("No tie.")
            print("Best matching entry:")
            print(top_matches[0]["question"])


# ---------------------------------------------------------
# MAIN FUNCTION
# ---------------------------------------------------------

def main():

    # Q1
    df = q1_create_knowledge_base()

    # Q2
    q2_scoring(df)

    # Q3
    q3_category(df)

    # Q4
    df = q4_add_keyword(df)

    # Q5
    q5_groupby(df)

    # Q6
    q6_tie_handling(df)


# Run program
if __name__ == "__main__":
    main()
```
