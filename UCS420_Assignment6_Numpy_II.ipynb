## Q1. Sensor Readings
Given temperature readings and an average measurement error of +2°C, perform the required operations without using a `for` loop.

```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
print("Original temperatures (°C):", temperature)
```

```python
# a. Add +2°C to every reading using vectorization
corrected_temperature = temperature + 2
print("Corrected temperatures (°C):", corrected_temperature)
```

```python
# b. Convert corrected temperatures from Celsius to Fahrenheit
fahrenheit = (9 / 5) * corrected_temperature + 32
print("Temperatures (°F):", fahrenheit)
```

```python
# c. Identify all corrected readings greater than 32°C using Boolean indexing
above_32 = corrected_temperature[corrected_temperature > 32]
print("Readings greater than 32°C:", above_32)
```

```python
# d. Count how many corrected readings exceed 32°C
count_above_32 = np.sum(corrected_temperature > 32)
print("Number of readings exceeding 32°C:", count_above_32)
```

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

print("Steps matrix:\n", steps)
```

```python
# a. Total steps recorded
total_steps = np.sum(steps)
print("Total steps:", total_steps)

# b. Mean number of steps
mean_steps = np.mean(steps)
print("Mean steps:", mean_steps)

# c. Maximum and minimum values
max_steps = np.max(steps)
min_steps = np.min(steps)
print("Maximum steps:", max_steps)
print("Minimum steps:", min_steps)

# d. Total steps for each day (axis=0)
daily_totals = np.sum(steps, axis=0)
print("Total steps for each day:", daily_totals)

# e. Total steps for each user (axis=1)
user_totals = np.sum(steps, axis=1)
print("Total steps for each user:", user_totals)

# f. User/day position containing the maximum number of steps
max_index = np.argmax(steps)
max_position = np.unravel_index(max_index, steps.shape)
print("Flat index of maximum:", max_index)
print("User/day position of maximum (0-based):", max_position)
print("Maximum value:", steps[max_position])
```

For Q2(f), `np.argmax()` first returns the index in the flattened array. `np.unravel_index()` converts that index back to the matrix's row-column position. The position is reported using zero-based indexing.

## Q3. NumPy Arrays, Slicing, Copying, Reshaping and Flattening

```python
# a. Create the original array
original = np.array([1, 2, 3, 4, 5, 6])
print("Original:", original)
```

```python
# b. Slice elements from index 1 to 4 (4 is excluded)
subset = original[1:5]
print("Subset:", subset)

# c. Modify the first element of subset
subset[0] = 999
print("After modifying subset:")
print("Original:", original)
print("Subset:", subset)
```

A basic NumPy slice is a **view**, not an independent copy. Therefore, changing `subset` also changes the corresponding element of `original`.

```python
# d. Create another slice using .copy() and modify it
copied_array = original[1:5].copy()
copied_array[0] = 500
print("Original after modifying copied array:", original)
print("Copied array:", copied_array)
```

```python
# e. Create numbers 1 to 12 and reshape into a 3 × 4 matrix
matrix = np.arange(1, 13).reshape(3, 4)
print("3 × 4 matrix:\n", matrix)
```

```python
# f. Indexing and slicing
first_row = matrix[0]
last_row = matrix[-1]
second_column = matrix[:, 1]
rows_1_2_cols_2_3 = matrix[1:3, 2:4]

print("First row:", first_row)
print("Last row:", last_row)
print("Second column:", second_column)
print("Rows 1–2 and columns 2–3 (using 1-based assignment wording):\n", rows_1_2_cols_2_3)
```

For the requested 3×4 matrix, the expression `matrix[1:3, 2:4]` selects rows 2–3 and columns 3–4 in 1-based human counting. NumPy itself uses zero-based indexing.

```python
# g. Flatten using both flatten() and ravel()
flattened = matrix.flatten()
raveled = matrix.ravel()

print("flatten():", flattened)
print("ravel():", raveled)
```

```python
# h. Modify an element returned by ravel() and observe the original array
raveled[0] = 1000
print("After modifying ravel():")
print("Original matrix:\n", matrix)
print("Raveled array:", raveled)
```

In this standard C-order case, `ravel()` returns a view when possible, so modifying it can affect the original array.

```python
# i. Modify an element returned by flatten() and observe the original array
flattened[1] = 2000
print("After modifying flatten():")
print("Original matrix:\n", matrix)
print("Flattened array:", flattened)
```

`flatten()` always returns a copy, so changing `flattened` does not change the original matrix.

```python
# j. Display shape, ndim, size and dtype
print("Shape:", matrix.shape)
print("Number of dimensions (ndim):", matrix.ndim)
print("Number of elements (size):", matrix.size)
print("Data type (dtype):", matrix.dtype)
```

## Q4. Cognitive-Assistive System and Ordinary Least Squares

Features: sleep hours, activity level, and stress level. Each row represents one user.

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

new_user = np.array([5, 40, 7])

print("X:\n", X)
print("y:", y)
print("New user:", new_user)
```

```python
# a. Display shape and dimensions of X
print("Shape of X:", X.shape)
print("Number of dimensions of X:", X.ndim)
```

```python
# b. Transpose X
XT = X.T
print("X.T:\n", XT)
```

### What does the transpose represent?
`X` has shape `(4, 3)`: four users and three features. `X.T` has shape `(3, 4)`, so each row of the transpose corresponds to one feature and contains that feature's values for all four users.

```python
# c. Calculate X.T @ X
XTX = X.T @ X
print("X.T @ X:\n", XTX)
```

```python
# d. Calculate the matrix inverse
# X.T @ X is a 3 × 3 square matrix. Check its determinant before inversion.
det_XTX = np.linalg.det(XTX)
print("det(X.T @ X):", det_XTX)

XTX_inv = np.linalg.inv(XTX)
print("(X.T @ X)^(-1):\n", XTX_inv)
```

```python
# e. Ordinary Least Squares: beta = (X.T X)^(-1) X.T y
beta = np.linalg.inv(X.T @ X) @ X.T @ y
print("OLS coefficients (beta):", beta)
```

```python
# g. Predict the assistance score for the new user
predicted_score = new_user @ beta
print("Predicted assistance score:", predicted_score)
```
