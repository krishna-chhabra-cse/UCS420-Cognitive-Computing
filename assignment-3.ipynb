```python
import pandas as pd


# =========================================================
# Q1. Create Dataset
# =========================================================

def q1_create_dataset():

    data = {
        "Employee_ID": [101, 102, 103, 104, 105],
        "Name": ["Alice", "Bob", "Charlie", "Diana", "Eve"],
        "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
        "Years_of_Experience": [2, 3, 4, 2, 3],
        "Age": [29, 34, 41, 28, 38],
        "Salary": [5000, 7000, 6500, 5500, 6000],
        "Joining_Date": [
            "2020-03-15",
            "2017-07-19",
            "2013-06-01",
            "2021-02-10",
            "2010-11-25"
        ],
        "Gender": ["Female", "Male", "Male", "Female", "Male"],
        "Bonus": [500, 700, 600, 400, 500],
        "Performance_Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
    }

    df = pd.DataFrame(data)

    print("\n========== Q1 ==========")
    print(df)

    return df
```

```python
# Q2. Locate rows 0, 4, 7 and 8

def q2_locate_rows(df):

    print("\n========== Q2 ==========")

    # The dataset created in Q1 has only rows 0 to 4.
    # Therefore rows 7 and 8 do not exist.

    valid_rows = [i for i in [0, 4, 7, 8] if i in df.index]

    print(df.loc[valid_rows])
```

```python
# Q3. Navigate DataFrame

def q3_navigation(df):

    print("\n========== Q3 ==========")

    # 1. Select row from index 3 to 7
    print("\n1. Rows from index 3 to 7:")
    print(df.loc[3:7])

    # 2. Select row 4 to 8 and column 2 to 4
    print("\n2. Rows 4 to 8 and columns 2 to 4:")

    # iloc uses positional indexes
    print(df.iloc[4:9, 2:5])

    # 3. All rows with column index 1 to 3
    # include column index 3
    print("\n3. All rows with column index 1 to 3:")
    print(df.iloc[:, 1:4])
```

```python
# Q4. Read CSV and display first 5 rows

def q4_read_csv():

    print("\n========== Q4 ==========")

    filename = "Iris.csv"

    try:
        df = pd.read_csv(filename)

        print("First five rows:")
        print(df.head())

        return df

    except FileNotFoundError:
        print(f"{filename} not found.")
        print("Download Iris.csv from the Kaggle dataset mentioned in the assignment.")

        return None


# =========================================================
# Q5. Delete row 4 and column 3
# =========================================================

def q5_delete(df):

    print("\n========== Q5 ==========")

    if df is None:
        return

    # Delete row 4
    df = df.drop(index=4)

    # Delete column at index 3
    column_to_delete = df.columns[3]
    df = df.drop(columns=[column_to_delete])

    print("After deleting row 4 and column 3:")
    print(df)


# =========================================================
# Q6. Employee Dataset
# =========================================================

def q6_create_employee_csv():

    data = {
        "Employee_ID": [101, 102, 103, 104, 105],
        "Name": ["Alice", "Bob", "Charlie", "Diana", "Eve"],
        "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
        "Years_of_Experience": [2, 3, 4, 2, 3],
        "Age": [29, 34, 41, 28, 38],
        "Salary": [5000, 7000, 6500, 5500, 6000],
        "Joining_Date": [
            "2020-03-15",
            "2017-07-19",
            "2013-06-01",
            "2021-02-10",
            "2010-11-25"
        ],
        "Gender": ["Female", "Male", "Male", "Female", "Male"],
        "Bonus": [500, 700, 600, 400, 500],
        "Performance_Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
    }

    df = pd.DataFrame(data)

    df.to_csv("employees.csv", index=False)

    return df


# =========================================================
# Q6(a). Shape
# =========================================================

def q6a_shape(df):

    print("\n========== Q6(a) ==========")

    print("Shape:", df.shape)


# =========================================================
# Q6(b). Summary / Info
# =========================================================

def q6b_summary(df):

    print("\n========== Q6(b) ==========")

    df.info()


# =========================================================
# Q6(c). Descriptive Statistics
# =========================================================

def q6c_descriptive_statistics(df):

    print("\n========== Q6(c) ==========")

    print(df.describe())


# =========================================================
# Q6(d). First 5 and Last 3 Rows
# =========================================================

def q6d_first_last(df):

    print("\n========== Q6(d) ==========")

    print("\nFirst 5 rows:")
    print(df.head(5))

    print("\nLast 3 rows:")
    print(df.tail(3))


# =========================================================
# Q6(e). Statistics
# =========================================================

def q6e_statistics(df):

    print("\n========== Q6(e) ==========")

    # i. Average salary
    average_salary = df["Salary"].mean()

    # ii. Total bonus
    total_bonus = df["Bonus"].sum()

    # iii. Youngest employee's age
    youngest_age = df["Age"].min()

    # iv. Highest performance rating
    highest_rating = df["Performance_Rating"].max()

    print("i. Average Salary:", average_salary)
    print("ii. Total Bonus:", total_bonus)
    print("iii. Youngest Employee's Age:", youngest_age)
    print("iv. Highest Performance Rating:", highest_rating)


# =========================================================
# Q6(f). Sort by Salary descending
# =========================================================

def q6f_sort_salary(df):

    print("\n========== Q6(f) ==========")

    sorted_df = df.sort_values(
        by="Salary",
        ascending=False
    )

    print(sorted_df)

    return sorted_df


# =========================================================
# Q6(g). Performance Category
# =========================================================

def q6g_performance_category(df):

    print("\n========== Q6(g) ==========")

    def categorize(rating):

        if rating >= 4.5:
            return "Excellent"

        elif rating >= 4.0:
            return "Good"

        else:
            return "Average"

    df["Performance_Category"] = (
        df["Performance_Rating"].apply(categorize)
    )

    print(df)

    return df


# =========================================================
# Q6(h). Identify Missing Values
# =========================================================

def q6h_missing_values(df):

    print("\n========== Q6(h) ==========")

    print(df.isnull().sum())


# =========================================================
# Q6(i). Rename Employee_ID to ID
# =========================================================

def q6i_rename(df):

    print("\n========== Q6(i) ==========")

    df = df.rename(
        columns={"Employee_ID": "ID"}
    )

    print(df)

    return df


# =========================================================
# Q6(j). Find Employees
# =========================================================

def q6j_find_employees(df):

    print("\n========== Q6(j) ==========")

    # i. More than 5 years experience
    print("\ni. Employees with more than 5 years experience:")

    experience = df[
        df["Years_of_Experience"] > 5
    ]

    print(experience)

    # ii. IT department
    print("\nii. Employees belonging to IT department:")

    it_employees = df[
        df["Department"] == "IT"
    ]

    print(it_employees)


# =========================================================
# Q6(k). Add Tax column
# =========================================================

def q6k_add_tax(df):

    print("\n========== Q6(k) ==========")

    # Tax = 10% of salary
    df["Tax"] = df["Salary"] * 0.10

    print(df)

    return df


# =========================================================
# Q6(l). Save Modified DataFrame
# =========================================================

def q6l_save(df):

    print("\n========== Q6(l) ==========")

    filename = "employees_modified.csv"

    df.to_csv(filename, index=False)

    print(f"Modified DataFrame saved to {filename}")


# =========================================================
# MAIN FUNCTION
# =========================================================

def main():

    # -------------------------
    # Q1
    # -------------------------
    df = q1_create_dataset()

    # -------------------------
    # Q2
    # -------------------------
    q2_locate_rows(df)

    # -------------------------
    # Q3
    # -------------------------
    q3_navigation(df)

    # -------------------------
    # Q4
    # -------------------------
    iris_df = q4_read_csv()

    # -------------------------
    # Q5
    # -------------------------
    q5_delete(iris_df)

    # -------------------------
    # Q6
    # -------------------------
    employee_df = q6_create_employee_csv()

    q6a_shape(employee_df)

    q6b_summary(employee_df)

    q6c_descriptive_statistics(employee_df)

    q6d_first_last(employee_df)

    q6e_statistics(employee_df)

    employee_df = q6f_sort_salary(employee_df)

    employee_df = q6g_performance_category(employee_df)

    q6h_missing_values(employee_df)

    employee_df = q6i_rename(employee_df)

    q6j_find_employees(employee_df)

    employee_df = q6k_add_tax(employee_df)

    q6l_save(employee_df)


# =========================================================
# RUN
# =========================================================

if __name__ == "__main__":
    main()
```
