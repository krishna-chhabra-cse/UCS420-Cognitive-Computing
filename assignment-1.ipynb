```python
import random


random.seed(1024170001)

nums = [random.randint(100, 900) for _ in range(100)]
print(nums)
```

```python
odd_nums = [x for x in nums if x % 2 != 0]
print("Odd numbers:", odd_nums)
print("Count:", len(odd_nums))
```

```python
even_nums = [x for x in nums if x % 2 == 0]
print("Even numbers:", even_nums)
print("Count:", len(even_nums))
```

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

prime_nums = [x for x in nums if is_prime(x)]

print("Prime numbers:", prime_nums)
print("Count:", len(prime_nums))
```

```python
from collections import Counter

freq = Counter(nums)
most_common = freq.most_common(1)[0]

print("Most frequent number:", most_common[0])
print("Occurrences:", most_common[1])
```

```python

digits = [1,0,2,4,1,7,0,0,1,7]

A = {d * 7 for d in digits}
B = {d * 9 for d in digits}

print("Set A:", A)
print("Set B:", B)
```

```python
print( A.union(B))
```

```python
print( A.intersection(B))
```

```python
print("A - B:", A.difference(B))
print("B - A:", B.difference(A))
```

```python
print( A.symmetric_difference(B))
```

```python
print("A is subset of B:", A.issubset(B))
print("B is superset of A:", B.issuperset(A))
```
