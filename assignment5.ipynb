```python
#Q.4 Write program to create an 1-D NumPy array named <<Your Name>> with evenly spaced 25 numbers from 10 to 100 using linspace(). Print the dimensions of the array, shape, total elements, the data type of each element
# and total number of bytes consumed by the array. Find the transpose of this array using reshape() attribute. Can we do the same with T attribute?
import numpy as np

# Create 1-D array with 25 evenly spaced numbers from 10 to 100
java = np.linspace(10, 100, 25)

print("Array:")
print(java)

print("\nDimensions:", java.ndim)
print("Shape:", java.shape)
print("Total elements:", java.size)
print("Data type:", java.dtype)
print("Total bytes:", java.nbytes)

# Transpose using reshape()
transpose_array = java.reshape(1, 25)
print("\nTranspose using reshape():")
print(transpose_array)

# Transpose using T attribute
print("\nTranspose using T:")
print(java.T)
```

```python
#Q5. Create a 2-D Array of three rows and four columns, named ucs420_<your_name>> with following values – 10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35. Compute the mean, median, max, min, unique elements.
# Reshape the array to four rows and three columns and name it as reshaped_ ucs420_<your_name>>. Resize the array to two rows and three columns and name it as resized_ ucs420_<your_name>>.
import numpy as np

# Create 2-D array with 3 rows and 4 columns
ucs420_java = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original Array:")
print(ucs420_java)

# Mean
print("\nMean:", np.mean(ucs420_java))

# Median
print("Median:", np.median(ucs420_java))

# Maximum
print("Maximum:", np.max(ucs420_java))

# Minimum
print("Minimum:", np.min(ucs420_java))

# Unique elements
print("Unique elements:", np.unique(ucs420_java))

# Reshape to 4 rows and 3 columns
reshaped_ucs420_java = ucs420_java.reshape(4, 3)

print("\nReshaped Array (4 x 3):")
print(reshaped_ucs420_java)

# Resize to 2 rows and 3 columns
resized_ucs420_java = np.resize(ucs420_java, (2, 3))

print("\nResized Array (2 x 3):")
print(resized_ucs420_java)
```
